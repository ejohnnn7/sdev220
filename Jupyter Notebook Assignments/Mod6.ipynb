```python
from datetime import date

# Get today's date
today = date.today()

# Convert date to string
today_string = today.isoformat()

# Write to file
with open("today.txt", "w") as file:
    file.write(today_string)

print("Date written to file:", today_string)
```

```python
# Read the date from the file
with open("today.txt", "r") as file:
    today_string = file.read()

print("Date read from file:", today_string)
```

```python
from datetime import datetime

# Parse string into date
parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

print("Parsed date:", parsed_date)
print("Type:", type(parsed_date))
```

```python
import multiprocessing
import random
import time
from datetime import datetime


# This function will run in each process
def wait_and_print():
    # Pick a random number between 0 and 1
    delay = random.random()

    # Wait that many seconds
    time.sleep(delay)

    # Print the current time
    print("Waited", round(delay, 3), "seconds.")
    print("Current time is:", datetime.now())
    print("Process finished.\n")


if __name__ == "__main__":
    # Create three processes
    process1 = multiprocessing.Process(target=wait_and_print)
    process2 = multiprocessing.Process(target=wait_and_print)
    process3 = multiprocessing.Process(target=wait_and_print)

    # Start the processes
    process1.start()
    process2.start()
    process3.start()

    # Wait for all processes to finish
    process1.join()
    process2.join()
    process3.join()

    print("All processes are done.")
```
